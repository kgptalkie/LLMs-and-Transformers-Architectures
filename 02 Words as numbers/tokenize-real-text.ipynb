```python
from transformers import AutoTokenizer
```

```python
tokenizer = AutoTokenizer.from_pretrained("openai-community/gpt2")
```

```python
len(tokenizer)
```

```python
text = "The cat sat because it was tired"
```

```python
ids = tokenizer.encode(text)
ids
```

```python
tokenizer.convert_ids_to_tokens(ids)
```

```python
tokenizer.decode(ids)
```

```python
tokenizer.tokenize("cat Cat cat")
```

```python
tokenizer.tokenize("The price is 12345 dollars")
```

```python
tokenizer.tokenize("for i in range(10):")
```

```python
len(tokenizer.encode("Hello world"))
```

```python
len(tokenizer.encode("Bonjour le monde"))
```

```python
len(tokenizer.encode("नमस्ते दुनिया"))
```
